```python
from sklearn.preprocessing import StandardScaler
dummies = ['Fuel_Type_CNG', 'Fuel_Type_Diesel',
       'Fuel_Type_Petrol', 'Color_Beige', 'Color_Black', 'Color_Blue',
       'Color_Green', 'Color_Grey', 'Color_Red', 'Color_Silver',
       'Color_Violet', 'Color_White', 'Color_Yellow']
features = ['Age_08_04','KM',
       'HP','Met_Color', 'Automatic', 'CC', 'Doors','Weight']+dummies

```

```python
x = df_toyota_dummies[features]
y = df_toyota_dummies['Price']
print(x.shape)
print(y.shape)
```
